```python
import requests
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt
from io import BytesIO

# Função para converter uma imagem colorida para tons de cinza
def converter_para_cinza(img):
    largura, altura = img.size
    img_cinza = Image.new("L", (largura, altura))
    for x in range(largura):
        for y in range(altura):
            r, g, b = img.getpixel((x, y))
            # Fórmula de luminância para conversão
            cinza = int(0.299 * r + 0.587 * g + 0.114 * b)
            img_cinza.putpixel((x, y), cinza)
    return img_cinza

# Função para binarizar uma imagem em tons de cinza
def binarizar_imagem(img, limiar=128):
    largura, altura = img.size
    img_binaria = Image.new("1", (largura, altura))
    for x in range(largura):
        for y in range(altura):
            pixel = img.getpixel((x, y))
            # Aplicação do limiar para binarização
            binario = 1 if pixel > limiar else 0
            img_binaria.putpixel((x, y), binario)
    return img_binaria
```

```python
# URL da imagem
url = "https://upload.wikimedia.org/wikipedia/en/7/7d/Lenna_%28test_image%29.png"

# Baixar e exibir a imagem original
response = requests.get(url)
img_original = Image.open(BytesIO(response.content))
plt.figure(figsize=(10, 5))
plt.subplot(1, 3, 1)
plt.title("Imagem Original")
plt.imshow(img_original)
plt.axis("off")
```

```python
# Converter a imagem para tons de cinza e exibir
img_cinza = converter_para_cinza(img_original)
plt.subplot(1, 3, 2)
plt.title("Imagem em Tons de Cinza")
plt.imshow(img_cinza, cmap="gray")
plt.axis("off")
```

```python
# Binarizar a imagem e exibir
img_binaria = binarizar_imagem(img_cinza)
plt.subplot(1, 3, 3)
plt.title("Imagem Binarizada")
plt.imshow(img_binaria, cmap="gray")
plt.axis("off")
```
